Sorting an array is  simple with `np.sort()`. we  can specify the axis , kind and order when we call the function .

```python
import numpy as np 
arr = np.array([2,6,4,1,8,19,45,39])
arr
```

```python
np.sort(arr) # It will sort the value in ascending order
```

```python
a = np.array(["a","b","d","c","f","z","p","r"])
print(np.sort(a))
print(type(a))
# p = []
# for i in a:
#     print((i))
#     p.append(i)
# print(p)
```

#### In additional to sort , which return a sorted copy of an array ,  we can use :

- `argsort`: Which is an indirect sort along a specified axis .
- `lexsort`: Which is an indirect stable sort on multiple keys .
- `searchsorted`: Which will find elements in a sorted array , 
- `partition`: which is a partial sort.

```python
ar = np.array([89,34,2,15,2,0,2,3,6])
print(np.argsort(ar))  # it sort the value with index wise for example 0 is located in  5th position so 5 is in first.
```

### We can concatenate with `np.cnocatenate()` function.

```python
a = np.array([1,2,3,4])
b = np.array([6,7,8,9])
print(np.concatenate((a, b)))
```

```python
# concatinate with 2d 
x = np.array([[1,2],[3,4]])
y = np.array([[5,6],[7,8]])
print(np.concat((x,y),axis = 0)) #row wise 
print(np.concat((x,y),axis = 1)) # column wise 
print(np.concatenate((x,y),axis=0)) # row wise and both are some 
```
