Sorting an array is  simple with `np.sort()`. we  can specify the axis , kind and order when we call the function .

```python
import numpy as np 
arr = np.array([2,6,4,1,8,19,45,39])
arr
```

```python
np.sort(arr) # It will sort the value in ascending order
```

```python
a = np.array(["a","b","d","c","f","z","p","r"])
print(np.sort(a))
print(type(a))
# p = []
# for i in a:
#     print((i))
#     p.append(i)
# print(p)
```

#### In additional to sort , which return a sorted copy of an array ,  we can use :

- `argsort`: Which is an indirect sort along a specified axis .
- `lexsort`: Which is an indirect stable sort on multiple keys .
- `searchsorted`: Which will find elements in a sorted array , 
- `partition`: which is a partial sort.

```python
ar = np.array([89,34,2,15,2,0,2,3,6])
print(np.argsort(ar))  # it sort the value with index wise for example 0 is located in  5th position so 5 is in first.
```

### We can concatenate with `np.cnocatenate()` function.

```python
a = np.array([1,2,3,4])
b = np.array([6,7,8,9])
print(np.concatenate((a, b)))
```

```python
# concatinate with 2d 
x = np.array([[1,2],[3,4]])
y = np.array([[5,6],[7,8]])
print(np.concat((x,y),axis = 0)) #row wise 
print(np.concat((x,y),axis = 1)) # column wise 
print(np.concatenate((x,y),axis=0)) # row wise and both are some 
```

In order to remove elements from an array , it's simple to use indexing to select the elements that you want to keep

```python
#Adding to element in the array 
import numpy as np 
# 1) np.append(array,value) - which add at the last of the array 
a = np.arange(1,30,2, dtype= np.int64)
new_array = np.append(a,30)
print(new_array)
print(a)
```

From the above code we can analysis , when we add an element in the array then it create new array rather than update the original

```python
# adding multiple elemenet in the array 
new_array = np.append(a,[20,40,60])
print(new_array)
```

```python
# adding an element in 2D array
arp = np.array([[1,2],[3,4],[5,6]])
arp
```

```python
new_array = np.append(arp,[[10,11]],axis = 0 ) # adding an element in row 
new_array
```

```python
np.append(arp,[[10],[21],[50]], axis = 1) # adding in column 
```

### Deleting  the element from the array by using 
- np.delete(array,index)

```python
arr = np.arange(2,10,2)
arr
```

```python
np.delete(arr,1)
```

```python
np.delete(arr,2)
```

```python
# delete multiple elemenet 

np.delete(arr,[1,2,3])
```

```python
# Deleteing in 2D 

arr = np.arange(1,17).reshape(4,4)
arr
```

```python
np.delete(arr,0,axis = 1) # it remove the  very first column 
```

```python
np.delete(arr,1,axis=0) # it remove the 2nd row 
```

```python
# deleteing multiple rows and column 
np.delete(arr,[0,3],axis = 1) # column 
```

```python
np.delete(arr,[1,3] , axis =0) # row 
```
