# Broadcasting :

There are times when you might want to carry out an operations between an array and a single number ( also called an Operation between a vectors and a scalar) or between arrays of two  different sizes. For example , the array ( called it data) might contains information about distance in miles but you want to convert the information to kilometers ,  We can perform this operation with :

```python
import numpy as np
data = np.array([1,2,3])
data * 1.6
```

![alt text](np_multiply_broadcasting.png)
Numpy Understands that the muttiplication should happen with each cell . That concept is called **Broadcasting** . Broadcasting is a mechanism that allow NumPy to perform operations on arrays of different shapes . The dimensions of the array must be compatible , For example When the dimensions are not compatible , we will get `**ValueError**`

```python
# 1) A number with an array :
# NumPy applies the number to every element :

a = np.array([10,20,30])
a+5
```

```python
# 2) Array with the same shape :
# Numpy operates element by element :
b = np.array([11,22,33])
a+b
```

```python
# 3) A row added to every rows :

a = np.array([[1,2,3],
              [4,5,6]])
b = np.array([10,20,30])
a+b
```

NumPy broadcasts `b` across both rows .
The compatibility rules :
Compare shape **From right to left** . Each pairs of dimensions must be :
- equal , or 
- one of them must be `1`

For example , Shapes (2,3) and (3,) are compatible : the trailing dimensions are both `3`, and the shorter shape is treated as (1,3). But (2,3) and (2) are not compatible , because the trailing dimensions `3` and `2` differ.
