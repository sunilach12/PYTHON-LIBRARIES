## How do you know the shape and size of an array ?
1) `ndarray.ndim`
2) `ndarray.size` 
3) `ndarray.shape`

#### `ndarray.ndim` : tell that the number of axes , or dimensions , of an array 
#### `ndarray.size` : tell that the total number of an element of the array , This is the product of the element of the arrays shape .
#### `ndarray.shape` :  display a tuple of integers that indicate the number of elements stored along each dimension of the array ,  for example   having 2D arrau with 2 rows and 3 column , the shape of your array is `(2,3)`.



```python
import numpy as np
a = np.array([[[0,1,2,3],
[4,5,6,7]],
[[0,1,2,3,],
[4,5,6,7]],
[[0,1,2,3],
[4,5,6,7]]])
# The number  of dimension of the array:
a.ndim
```

```python
# The total number of elements in the array ,
a.size
```

```python
# the shape of the array 
a.shape
```

# Can we reshape an array ?
Yes ! 
by using the `arr.reshape()` will return a reshaped array without changing the data . Just remember that when we use the reshape method , the array we want to produced  needs to have the smae number of elements as the original array . if we start with an array with 12 elements , we will need to make sure that our new array also has a total  of 12 elements.


```python
a = np.arange(6)
print(a)
```

#### now we can `reshape()` to reshape the array . For example you can reshape this array to an array with three rows and two column , and vice versa.

```python
b = a.reshape(2,3)
print(b)
c =a.reshape(3,2)
print(c)
```

```python
np.reshape(a,shape=(2,3))
```
