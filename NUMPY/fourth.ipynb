## How to convert a 1D array into a 2D array (How to add a new axis to an array)

`np.newaxis` , `np.expand_dims`

we can use `np.newaxis` and `np.expand_dims` to increase the dimensions of existing array.
Using `np.newaxis` will increase the dimensions of your array by one dimension when used once . This means that a ***1D*** array  will becomes a **2D** array, a **2D** will becomes a **3D** array , and so on 

For example 

```python
import numpy as np
a = np.array([1,2,3,4,5,6])
a.shape
```

```python
a2 = a[np.newaxis,:]
print(a2.shape)
print(a2)
```

```python
a3 = a[:,np.newaxis]
print(a3)
a3.shape
```

You can explicitly convert a 1D array to either a row vector or a column vector using `np.newaxis`. . For example  , We can convert a 1D to a row vector by inserting an axis along the first dimension.


```python
row = a[np.newaxis,:]
row.shape
```

For a column vectors , we can use insert an axis along the second  dimension .

```python
col = a[:,np.newaxis]
col.shape
```

also expand an array by inserting a new axis at a specified position with `np.expand_dims`

```python
b = np.expand_dims(a,axis = 1)
b.shape
```

```python
c = np.expand_dims(a,axis=1)
c.shape
```
