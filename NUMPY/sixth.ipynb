Basic operation are simple with numpy , if find the sum of the element in an array ,  we had use `sum()` . this works for 1D array , 2D array and arrays in the higher dimensions

```python
a = np.array([1,2,3,4])
total = a.sum()
print(total)
print(total.dtype)
```

Toa add the rows or the columns in a 2D array , we would specify the axis .

```python
b = np.array([[1,1],[2,2]])
sum_in_rows = b.sum(axis = 0)
sum_in_rows
```

```python
sum_in_column = b.sum(axis = 1)
sum_in_column
```
