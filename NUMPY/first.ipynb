Often the element of an array are originally unknown , but its size is known . Hence ,Numpy offer several functions to create arrays with initial placeholder content . These minimize the necessity of growing array , an expensive operation . 
- The function Zeros Create an array full of zeros ,
- the function ones creates an array full of ones , 
- And the function empty create an array whose initial content is random and depends on the state of the memory , By the default , ``the dtype of the created array is float64.``

```python
np.zeros((3,4))
```

```python
np.ones((3,4), dtype = np.int16)
```

## **How to Create a Basic array**
##### **The ways to create an array are given below :**
1) `np.zeros()`
2) `np.ones()`
3) `np.empty()`
4) `np.arange()`
5) `np.linspace()`

```python
# 1) np.zeros()

a = np.zeros((5,2)) # five rows and 2 column 
a
```

```python
b = np.zeros((3,4,5)) # 3 Blocks , 4Rows and 5 columns 
b
```

```python
b.size
```

```python
b.ndim
```

```python
b.shape
```

##### 2) np.ones(())

```python
c = np.ones((2,3,4)) #Same as the zeros  , it create one  instead of zero 
c
```

#### 3) np.empty()

Or even an empty array ! The function `empty` creates an array whose initial content is random and depends on the state of the memory , The reason to use `empty` over `zeros` ( Or something similar) is speed - just make sure to fill every element afterward !

```python
# Creating an empty array with 2 element 

d = np.empty(4)
d
```

##### 4) np.arange() 
---> creating an array with a range of elements : 

```python
e = np.arange(4)
e
```

And even an array that contained a range  of evenly spaced intervals . To do this , You will specify the **First Number , Last number  and the step size** like a regular python 

```python
f = np.arange(2,9,2)
f
```

##### You can also use **np.linspace()** to create  an array with the valuse that are spaced linearly in a specified interval:

```python
g = np.linspace(0,10, num=6) # num = 6 represent  to devide the number .
g
```

#### While the default datatype is floating point (`np.float64`), you can explicitly specify whihc data type you want using the `dtype` keyword

```python
x = np.ones((3,5),dtype = np.int64)
x
```
